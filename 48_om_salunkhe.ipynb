```python
def longest_increasing_subsequence(nums):
    if not nums:
        return []
    
    n = len(nums)
    dis = [1] * n
    t = [-1] * n
    
    for i in range(n):
        for j in range(i):
            if nums[i]>nums[j] and dis[j] + 1>dis[i]:
                dis[i] = dis[j] + 1
                t[i] = j
    max_len = max(dis)
    end_pos = dis.index(max_len)
    lis=[]
    current=end_pos
    while current!=-1:
        lis.append(nums[current])
        current=t[current]
    
    return lis[::-1]

def count_longest_subsequences(nums):
    if not nums:
        return 0
    
    n=len(nums)
    dis=[1]*n
    count=[1]*n
    
    for i in range(n):
        for j in range(i):
            if nums[i]>nums[j]:
                if dis[j]+1>dis[i]:
                    dis[i]=dis[j] + 1
                    count[i]=count[j]
                elif dis[j]+1==dis[i]:
                    count[i]+=count[j]
    
    max_len=max(dis)
    total=0
    
    for i in range(n):
        if dis[i]==max_len:
            total+=count[i]
    
    return total

inp=[10, 9, 2, 5, 3, 7, 101, 18]
lis=longest_increasing_subsequence(inp)

print(len(lis))
print(count_longest_subsequences(inp))
print(lis)
```
